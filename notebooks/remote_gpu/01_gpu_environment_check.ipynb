### Step 2: Detailed GPU Hardware Properties

```python
import subprocess

print("=== NVIDIA SMI DETAILED QUERY ===")
cmd = ["nvidia-smi", "--query-gpu=name,driver_version,memory.total,memory.free,compute_cap", "--format=csv,noheader"]
try:
    res = subprocess.run(cmd, capture_output=True, text=True, check=True)
    print("GPU Details:", res.stdout.strip())
except Exception as e:
    print("Failed to query nvidia-smi:", e)
```

### Step 3: PyTorch CUDA & cuDNN Environment Verification

```python
import sys, platform
import torch

print("=== PYTORCH & CUDA RUNTIME DIAGNOSTIC ===")
print(f"Python Version: {platform.python_version()}")
print(f"PyTorch Version: {torch.__version__}")
print(f"CUDA Available in PyTorch: {torch.cuda.is_available()}")
if torch.cuda.is_available():
    print(f"Device Count: {torch.cuda.device_count()}")
    print(f"Device Name: {torch.cuda.get_device_name(0)}")
    print(f"Device Capability: {torch.cuda.get_device_capability(0)}")
    print(f"CUDA Arch List: {torch.cuda.get_arch_list()}")
    print(f"CUDA Version (PyTorch built with): {torch.version.cuda}")
    print(f"cuDNN Enabled: {torch.backends.cudnn.enabled}")
    print(f"cuDNN Version: {torch.backends.cudnn.version()}")
    
    # Quick tensor allocation test
    x = torch.randn(1000, 1000, device='cuda')
    y = torch.matmul(x, x)
    torch.cuda.synchronize()
    print("CUDA Tensor Matrix Multiplication Test: SUCCESS")
else:
    print("WARNING: CUDA is NOT available in this PyTorch session!")
```

### Step 4: Export Diagnostic Evidence to File

```python
import json

diag_info = {
    "python_version": platform.python_version(),
    "pytorch_version": torch.__version__,
    "cuda_available": torch.cuda.is_available(),
    "cuda_version": torch.version.cuda if torch.cuda.is_available() else None,
    "cudnn_version": torch.backends.cudnn.version() if torch.cuda.is_available() else None,
    "device_name": torch.cuda.get_device_name(0) if torch.cuda.is_available() else "None",
    "compute_capability": torch.cuda.get_device_capability(0) if torch.cuda.is_available() else None
}

with open("gpu_environment_diagnostic.json", "w") as f:
    json.dump(diag_info, f, indent=2)

print("Diagnostic report saved to gpu_environment_diagnostic.json")
```
